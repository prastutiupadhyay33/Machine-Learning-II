**The below functions need to be defined to carry out the EDA.**

```python
def histogram_boxplot(data, feature, figsize=(15, 10), kde=False, bins=None):
    """
    Boxplot and histogram combined

    data: dataframe
    feature: dataframe column
    figsize: size of figure (default (15,10))
    kde: whether to show the density curve (default False)
    bins: number of bins for histogram (default None)
    """
    f2, (ax_box2, ax_hist2) = plt.subplots(
        nrows=2,  # Number of rows of the subplot grid= 2
        sharex=True,  # x-axis will be shared among all subplots
        gridspec_kw={"height_ratios": (0.25, 0.75)},
        figsize=figsize,
    )  # creating the 2 subplots
    sns.boxplot(
        data=data, x=feature, ax=ax_box2, showmeans=True, color="violet"
    )  # boxplot will be created and a triangle will indicate the mean value of the column
    sns.histplot(
        data=data, x=feature, kde=kde, ax=ax_hist2, bins=bins
    ) if bins else sns.histplot(
        data=data, x=feature, kde=kde, ax=ax_hist2
    )  # For histogram
    ax_hist2.axvline(
        data[feature].mean(), color="green", linestyle="--"
    )  # Add mean to the histogram
    ax_hist2.axvline(
        data[feature].median(), color="black", linestyle="-"
    )  # Add median to the histogram
```

```python
# function to create labeled barplots


def labeled_barplot(data, feature, perc=False, n=None):
    """
    Barplot with percentage at the top

    data: dataframe
    feature: dataframe column
    perc: whether to display percentages instead of count (default is False)
    n: displays the top n category levels (default is None, i.e., display all levels)
    """

    total = len(data[feature])  # length of the column
    count = data[feature].nunique()
    if n is None:
        plt.figure(figsize=(count + 2, 6))
    else:
        plt.figure(figsize=(n + 2, 6))

    plt.xticks(rotation=90, fontsize=15)
    ax = sns.countplot(
        data=data,
        x=feature,
        palette="Paired",
        order=data[feature].value_counts().index[:n],
    )

    for p in ax.patches:
        if perc == True:
            label = "{:.1f}%".format(
                100 * p.get_height() / total
            )  # percentage of each class of the category
        else:
            label = p.get_height()  # count of each level of the category

        x = p.get_x() + p.get_width() / 2  # width of the plot
        y = p.get_height()  # height of the plot

        ax.annotate(
            label,
            (x, y),
            ha="center",
            va="center",
            size=12,
            xytext=(0, 5),
            textcoords="offset points",
        )  # annotate the percentage

    plt.show()  # show the plot
```

## Bagging - Model Building and Hyperparameter Tuning

- We'll fit different models on the train data and observe their performance.
- We'll try to improve that performance by tuning some hyperparameters available for that algorithm.
- We'll use GridSearchCv for hyperparameter tuning and `r_2 score` to optimize the model.
- R-square - `Coefficient of determination` is used to evaluate the performance of a regression model. It is the amount of the variation in the output dependent attribute which is predictable from the input independent variables.
- Let's start by creating a function to get model scores, so that we don't have to use the same codes repeatedly.

```python
# Function to compute adjusted R-squared
def adj_r2_score(predictors, targets, predictions):
    r2 = r2_score(targets, predictions)
    n = predictors.shape[0]
    k = predictors.shape[1]
    if n <= k + 1:
        return np.nan
    return 1 - ((1 - r2) * (n - 1) / (n - k - 1))


# Function to compute MAPE, ignoring zero target values to avoid division by zero
def mape_score(targets, predictions):
    targets = np.asarray(targets)
    predictions = np.asarray(predictions)
    nonzero = targets != 0
    if not np.any(nonzero):
        return np.nan
    return np.mean(np.abs((targets[nonzero] - predictions[nonzero]) / targets[nonzero])) * 100


# Function to compute regression model performance
def model_performance_regression(model, predictors, target):
    pred = model.predict(predictors)

    r2 = r2_score(target, pred)
    adjr2 = adj_r2_score(predictors, target, pred)
    rmse = np.sqrt(mean_squared_error(target, pred))
    mae = mean_absolute_error(target, pred)
    mape = mape_score(target, pred)

    return pd.DataFrame(
        {
            "RMSE": [rmse],
            "MAE": [mae],
            "R-squared": [r2],
            "Adj. R-squared": [adjr2],
            "MAPE": [mape],
        }
    )
```

```python
##  Function to calculate r2_score and RMSE on train and test data
def get_model_score(model, flag=True):
    '''
    model : classifier to predict values of X

    '''
    # defining an empty list to store train and test results
    score_list=[]

    pred_train = model.predict(X_train)
    pred_test = model.predict(X_test)

    train_r2=metrics.r2_score(y_train,pred_train)
    test_r2=metrics.r2_score(y_test,pred_test)
    train_rmse=np.sqrt(metrics.mean_squared_error(y_train,pred_train))
    test_rmse=np.sqrt(metrics.mean_squared_error(y_test,pred_test))

    #Adding all scores in the list
    score_list.extend((train_r2,test_r2,train_rmse,test_rmse))

    # If the flag is set to True then only the following print statements will be dispayed, the default value is True
    if flag==True:
        print("R-sqaure on training set : ",metrics.r2_score(y_train,pred_train))
        print("R-square on test set : ",metrics.r2_score(y_test,pred_test))
        print("RMSE on training set : ",np.sqrt(metrics.mean_squared_error(y_train,pred_train)))
        print("RMSE on test set : ",np.sqrt(metrics.mean_squared_error(y_test,pred_test)))

    # returning the list with train and test scores
    return score_list
```

```python
# Training performance (tuned)
dtree_tuned_model_train_perf = model_performance_regression(dtree_tuned, X_train,y_train)
print("Training performance \n",dtree_tuned_model_train_perf)
```

### Hyperparameter Tuning

```python
# Choose the type of classifier.
xgb_tuned = XGBRegressor(random_state=1, verbosity = 0)

# Grid of parameters to choose from
parameters = {'n_estimators': [75,100,125,150],
              'subsample':[0.7, 0.8, 0.9, 1],
              'gamma':[0, 1, 3, 5],
              'colsample_bytree':[0.7, 0.8, 0.9, 1],
              'colsample_bylevel':[0.7, 0.8, 0.9, 1]
              }

# Type of scoring used to compare parameter combinations
scorer = metrics.make_scorer(metrics.r2_score)

# Run the grid search
grid_obj = GridSearchCV(xgb_tuned, parameters, scoring=scorer,cv=5,n_jobs=-1)
grid_obj = grid_obj.fit(X_train, y_train)

# Set the clf to the best combination of parameters
xgb_tuned = grid_obj.best_estimator_

# Fit the best algorithm to the data.
xgb_tuned.fit(X_train, y_train)
```

## Stacking - Model Building

- Stacking combines the predictions of several models and a final model learns how to blend them.
- We stack the tuned Random Forest, Gradient Boosting and XGBoost, with Linear Regression as the final model.

```python
# Stack the three best tuned models; a linear regression blends their predictions
stacking_estimator = StackingRegressor(
    estimators=[('rf', rf_tuned), ('gb', gb_tuned), ('xgb', xgb_tuned)],
    final_estimator=LinearRegression(),
    cv=5, n_jobs=-1)
stacking_estimator.fit(X_train, y_train)
```

```python
# Training performance
stacking_model_train_perf = model_performance_regression(stacking_estimator, X_train, y_train)
print("Training performance \n", stacking_model_train_perf)
```

```python
# Testing performance
stacking_model_test_perf = model_performance_regression(stacking_estimator, X_test, y_test)
print("Testing performance \n", stacking_model_test_perf)
```

- Stacking gives the best test score in our run (R2 approx 0.957, RMSE approx 38.5), slightly better than the best single model.
- The gain over tuned Gradient Boosting / XGBoost is small, so a single boosted model is also a good, simpler choice.

## Conclusions and Business Recommendations

### Conclusions

1. **Bike demand varies by time and season.** The exploratory analysis shows that demand changes across hours, months, seasons, and weather conditions. The exact pattern should be confirmed from the plots generated by running the notebook.
2. **Time of day is important.** Average rentals tend to be higher during commuting periods, particularly around morning and evening peak hours.
3. **Weather affects rentals.** Clear or partly cloudy conditions generally show higher average rentals than poor-weather conditions.
4. **The year and season features help explain demand.** The analysis compares rental counts across years and seasons to identify changes in usage.
5. **Ensemble models are useful for prediction.** The notebook evaluates Decision Tree, Random Forest, AdaBoost, Gradient Boosting, XGBoost, and Stacking regression models. Based on the model-comparison notes above, tuned Gradient Boosting performed strongly; Stacking was also reported to achieve a test R² of about 0.957 and RMSE of about 38.5 in that run. These figures can vary with library versions, data, and tuning settings, so use the freshly generated comparison table as the final authority.

### Business Recommendations

1. **Plan fleet availability around demand.** Keep more bikes available during observed peak hours and in seasons with higher average rentals, while avoiding overstocking during low-demand periods.
2. **Use weather forecasts in daily planning.** Adjust bike distribution and staffing based on expected weather, since poor weather is associated with lower demand.
3. **Place stations near commute destinations.** Consider high-demand residential areas, offices, colleges, transit points, and other locations supported by local usage data.
4. **Schedule maintenance during low-demand hours.** Where operationally practical, use quieter nighttime periods for routine maintenance and redistribution.
5. **Test promotions carefully.** Try subscriptions or targeted offers on lower-demand days or periods, then measure whether they increase rentals and revenue.
6. **Use predictions as decision support, not certainty.** Compare forecasts with actual rentals regularly and retrain the model when demand patterns change. Evaluate business outcomes as well as R² and RMSE before deploying a model.
7. **Avoid unsupported seasonal pricing decisions.** Consider demand-based pricing only after checking local rules, customer response, and the actual seasonal results in the analysis.

### Limitations and Next Steps

- The model uses historical hourly data, so unusual events, new stations, service disruptions, and changes in customer behavior may not be captured.
- The train/test split is shuffled. For a real forecasting deployment, also evaluate with a chronological split so that future observations are not used to predict the past.
- Compare the models using the generated test-performance table and validate the selected model on newer data before operational use.
